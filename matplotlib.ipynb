```python
fig, axs = plt.subplot_mosaic([['linear','linear-log'],
                              ['log-linear','log-log']], layout='constrained')
#plt.title('axes-scales')
x=np.arange(0,3 *np.pi,0.1)
y=2* np.sin(x) + 3

def plotter(title,axs=axs, x=x,y=y, xscale='linear', yscale='linear', xlabel='', ylabel=''):
    ax = axs[title]
    ax.plot(x,y)
    ax.set_xscale(xscale)
    ax.set_yscale(yscale)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    return axs

plotter('linear', xlabel='linear', ylabel='linear')
plotter('linear-log',yscale ='log', xlabel='linear', ylabel='log')
plotter('log-linear',xscale='log', xlabel='log', ylabel='linear')
plotter('log-log', xscale='log', yscale='log', xlabel='log', ylabel='log')
```

```python
from matplotlib.colors import LogNorm

X, Y = np.meshgrid(np.linspace(-3,3, 128), np.linspace(-3,3,128))
Z= (1-X/2 + X**5 + Y**3) * np.exp(-X**2 -Y**2)

fig,axs = plt.subplots(2,2, layout='constrained')
pc = axs[0,0].pcolormesh(X,Y, Z, vmin=-0.5, vmax=1, cmap='RdBu_r')
fig.colorbar(pc, ax=axs[0,0])

co = axs[0,1].contourf(X,Y,Z, levels=np.linspace(-1.25,1.25,11))
fig.colorbar(co, ax=axs[0,1])

pc = axs[1,0].imshow(Z**2 * 100, cmap='plasma', norm=LogNorm(vmin=0.01, vmax=100))
fig.colorbar(pc, ax=axs[1,0], extend='both')

pc = axs[1,1].scatter(data1, data2, c=data3, cmap='RdBu_r')
fig.colorbar(pc, ax=axs[1,1], extend='both')
```

```python
import matplotlib.animation as animation
fig, ax = plt.subplots()
t = np.linspace(0, 3, 40)
g = -9.81
v0 = 12
z = g * t**2 / 2 + v0 * t

v02 = 5
z2 = g * t**2 / 2 + v02 * t

scat = ax.scatter(t[0], z[0], c="b", s=5, label=f'v0 = {v0} m/s')
line2 = ax.plot(t[0], z2[0], label=f'v0 = {v02} m/s')[0]
ax.set(xlim=(0, 3), ylim=(-4, 10), xlabel='Time [s]', ylabel='Z [m]')
ax.legend()


def update(frame):
    # for each frame, update the data stored on each artist.
    x = t[:frame]
    y = z[:frame]
    # update the scatter plot:
    data = np.stack([x, y]).T
    scat.set_offsets(data)
    # update the line plot:
    line2.set_xdata(t[:frame])
    line2.set_ydata(z2[:frame])
    return (scat, line2)


ani = animation.FuncAnimation(fig=fig, func=update, frames=40, interval=30)
plt.show()
```

```python
from sklearn.datasets import fetch_openml
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

mnist=fetch_openml('mnist_784', as_frame=False)
X,y = mnist['data'], mnist['target']

X_train, X_test, X_valid = X[:50_000], X[50_000:60_000], X[60_000:]
y_train, y_test, y_valid = y[:50_000], y[50_000:60_000], y[60_000:]


from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import ExtraTreeClassifier
from sklearn.svm import SVC
```

```python
rfcl = RandomForestClassifier( random_state=42)
etcl = ExtraTreeClassifier(  random_state=42)
svccl= SVC(random_state=42)

rfcl.fit(X_train, y_train)
etcl.fit(X_train, y_train)
svccl.fit(X_train, y_train)
```

```python
print(rfcl.score(X_valid, y_valid))
print(etcl.score(X_valid,y_valid))
print(svccl.score(X_valid,y_valid))
```

```python
from sklearn.calibration import CalibratedClassifierCV
csvc = CalibratedClassifierCV(SVC(random_state=42))
from sklearn.ensemble import VotingClassifier

hard_clf = VotingClassifier(
    estimators =[
        ('rf' , RandomForestClassifier(random_state=42)),
        ('et', ExtraTreeClassifier(random_state=42)),
        ('svc', csvc)
    ]
)
hard_clf.fit(X_train, y_train)
         
```

```python
hard_clf.score(X_valid, y_valid)
```

```python
hard_clf.voting='soft'
#hard_clf.estimators[2] = CalibratedClassifierCV(SVC(random_state=42), cv=5)
#hard_clf.fit(X_train,y_train)

hard_clf.score(X_valid, y_valid)
```

```python
hard_clf.voting='hard'
et_trained = hard_clf.named_estimators_.pop('et')
hard_clf.estimators_.remove(et_trained)
hard_clf.score(X_valid, y_valid)
```

```python
import pandas as pd
import polars as pl
import datetime as dt

df = pl.DataFrame(
    { "name" : ["Alice","Bem", "Cholor", "Daniel"],
         "birthdate": [ 
             dt.date(1997,6,10),
             dt.date(1985,4,28),
             dt.date(1955,3,1),
             dt.date(2007,11,16)
         ],
     "weight": [57.9, 72.5, 53.6, 83.1],
     "height": [1.56,1.77, 1.93, 1.75],
    }
)
print(df.head())
df.write_csv("test_polars.csv")
result = df.select(pl.col("name"), pl.col("birthdate")
            .dt.month().alias("birth_month"), (pl.col("weight")/pl.col("height") **2).alias("bmi"))
result = result.filter(pl.col("birth_month") > 3)
print(result)

result = df.group_by(  (pl.col("birthdate").dt.year() // 10 * 10).alias("decade"), maintain_order=True).agg(pl.col("weight").sum())
print(result)
```

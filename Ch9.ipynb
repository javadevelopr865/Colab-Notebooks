```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.linear_model import Perceptron
import matplotlib.pyplot as plt
```

```python
iris = load_iris(as_frame=True)
X= iris.data[["petal length (cm)", "petal width (cm)"]].values
y = (iris.target == 0)

per_clf = Perceptron(random_state=42)
per_clf.fit(X,y)
```

```python
X_new = [[2,0.5], [3,1]]
y_pred = per_clf.predict(X_new)
y_pred
```

```python
from sklearn.datasets import fetch_california_housing
from sklearn.metrics import root_mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
```

```python
housing = fetch_california_housing()
X_train, X_test, y_train, y_test = train_test_split(housing.data, housing.target, random_state=42)
```

```python
mlp_reg = MLPRegressor(hidden_layer_sizes=[50,50,50], early_stopping=True, verbose=True, random_state=42)
pipeline = make_pipeline(StandardScaler(), mlp_reg)
pipeline.fit(X_train, y_train)
```

```python
mlp_reg.best_validation_score_
```

```python
y_pred = pipeline.predict(X_test)
```

```python
rmse = root_mean_squared_error(y_pred, y_test)
rmse
```

```python
plt.figure(figsize=(6,6))
plt.scatter(y_test, y_pred, s=5, alpha=0.3)
plt.plot([0,5.2],[0,5.2], color='red', linestyle='--', label='ideal predictions = targets')
plt.axis("equal")
plt.xlabel('Target')
plt.ylabel('Prediction')
plt.grid()
plt.legend(fontsize=12)
```



#### Typical architecture of regression MLP

| Hyperparameter | Typical value | 
| :---     | :----   |
| #hidden layers     | 1 to 5   | 
| #neurons per hidden layer     | typically 10 to 100     | 
| #output neurons | 1 per target dimension|
| Hidden layers activation | ReLU |
| Output activation | None or ReLU/Softplus (if positive) or sigmoid/tanh (if bounded)|
| Loss function | MSE, MAE or Huber (hybrid) if outliers|


## Classification MLPs

You just need a single output neuron using the sigmoid activation function: the output will be a number between 0 and 1 which we will
interprete as the estimated probability of the positive class.  

For multilabel binary classification you would need multiple output neurons  
For multinomial (i.e multiclass) classification we use the softmax activation for the whole output layer  

For the loss function we ditch MSE/MAE and use cross-entropy loss

The `MLPClassifier` class uses softmax activation for its output layer and minimizes the cross-entropy loss. The `score()` function returns the $R^2$ score


```python
from sklearn.datasets import fetch_openml

fashion_mnist = fetch_openml(name="Fashion-MNIST", as_frame=False)
targets = fashion_mnist.target.astype(int)
```

```python
fashion_mnist.data.shape
```

```python
x = fashion_mnist.data[18].reshape(28,28)
plt.figure(figsize=(0.5,0.5))
plt.imshow(x, cmap="binary")
plt.axis("off")
```

```python
X_train, y_train = fashion_mnist.data[:60_000], targets[:60_000]
X_test, y_test = fashion_mnist.data[60_000:], targets[60_000:]

class_names = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
               "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]
```

```python
class_names[y_train[18]]
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import MinMaxScaler

mlp_clf= MLPClassifier(hidden_layer_sizes =[300,100], verbose=True, 
                       early_stopping=True, random_state=42)

pipeline = make_pipeline(MinMaxScaler(), mlp_clf)
pipeline.fit(X_train, y_train)
accuracy = pipeline.score(X_test, y_test)
```

```python
mlp_clf.best_validation_score_
```

### Hyperparameter tuning
- Number of hidden layers: Deep networks have much higher parameter efficiency
- Number of Neurons per hidden layer: Can sometimes help to make the first hidden layer a bit larger than the others. Start with slightly more layers and neurons than you need and use early stopping and regularization to prevent it from overfitting. In general you get more bang for your buck by increasing the number of layers rather than the number of neurons per layer
- Learning Rate: One way to find a good lr is to train the model for a few hundred iterations starting with a very low rat e.g $10^-5$ and gradually increasing up to a very large value (e.g 10) by multiplying by a constant factor at each iteration
- Batch size:
- Optimizer: Choosing something other than mini-batch GD can help speed up training and sometimes reach better performance
- Activation function: In general ReLU  
_Make sure to tune the learning rate again if you modify any other hyperparemeter_
